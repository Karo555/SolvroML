```python
import pandas as pd
import json

# Load the JSON data
with open('data/Methotrexate__MTX_.json', 'r') as f:
    data = json.load(f)

# Create nodes DataFrame
nodes_data = []
for node_name, node_info in data['graph_data']['nodes'].items():
    nodes_data.append({
        'node_id': node_name,
        'source_count': len(node_info['sources'])
    })
    
nodes_df = pd.DataFrame(nodes_data)

# Create edges DataFrame
edges_data = []
for edge in data['graph_data']['edges']:
    edges_data.append({
        'source': edge['source'],
        'target': edge['target'],
        'relation': edge['relation'],
        'source_count': len(edge['sources'])
    })
    
edges_df = pd.DataFrame(edges_data)

# Create path DataFrame
path_data = []
for i in range(len(data['path_edges'])):
    path_data.append({
        'step': i+1,
        'source': data['path_edges'][i]['source'],
        'relation': data['path_edges'][i]['relation'],
        'target': data['path_edges'][i]['target']
    })
    
path_df = pd.DataFrame(path_data)

# Print the DataFrames
# print("Nodes DataFrame:")
# print(nodes_df)

# print("\nEdges DataFrame:")
# print(edges_df)

print("\nPath DataFrame:")
print(path_df)
```
